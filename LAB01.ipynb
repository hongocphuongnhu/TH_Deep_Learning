**TIỀN XỬ LÝ DỮ LIỆU CHO DATASET dulieuxettuyendaihoc.csv**

```python
import pandas as pd
dulieuxettuyen = pd.read_csv("https://raw.githubusercontent.com/huynhhoc/DataAnalystDeepLearning/main/Data/dulieuxettuyendaihoc.csv")
print (dulieuxettuyen.head(10))
print (dulieuxettuyen.tail(10))
```

```python
dulieuxettuyen.isnull().sum()
```

```python
dropAllNaNFrame = dulieuxettuyen.dropna()
print (dropAllNaNFrame)
```

```python
#thay thế các giá trị null thành số 0
dulieuxettuyen = dulieuxettuyen.fillna(0)
```

```python
# Khảo sát dữ liệu thiếu cho cột 'DT' (Dân tộc)
print("Các giá trị riêng biệt của DT:", dulieuxettuyen['DT'].unique())
tan_so_dt = dulieuxettuyen['DT'].value_counts(dropna=False)
tan_suat_dt = dulieuxettuyen['DT'].value_counts(dropna=False, normalize=True) * 100

thong_ke_dt = pd.DataFrame({'Tần số': tan_so_dt, 'Tần suất (%)': tan_suat_dt})
print("\nBảng thống kê DT:")
display(thong_ke_dt)

dulieuxettuyen['DT'] = dulieuxettuyen['DT'].fillna(0)
```

```python
#Thống kê dữ liệu thiếu cho T1 và các cột điểm khác, sau đó điền bằng Mean
columns_to_fix = ['T1', 'L1', 'H1', 'S1', 'V1', 'X1', 'D1', 'N1',
                  'T2', 'L2', 'H2', 'S2', 'V2', 'X2', 'D2', 'N2',
                  'T3', 'L3', 'H3', 'S3', 'V3', 'X3', 'D3', 'N3',
                  'T4', 'L4', 'H4', 'S4', 'V4', 'X4', 'D4', 'N4',
                  'T5', 'L5', 'H5', 'S5', 'V5', 'X5', 'D5', 'N5',
                  'T6', 'L6', 'H6', 'S6', 'V6', 'X6', 'D6', 'N6']

for col in columns_to_fix:
    if col in dulieuxettuyen.columns:
        mean_val = dulieuxettuyen[col].mean()
        # dulieuxettuyen[col] = dulieuxettuyen[col].replace(0, np.nan) # Trả lại nan vì trước đó điền 0
        dulieuxettuyen[col] = dulieuxettuyen[col].fillna(mean_val)

print("Đã xử lý xong dữ liệu thiếu cho các biến điểm số bằng phương pháp Mean.")
print("Kiểm tra lại dữ liệu thiếu:")
print(dulieuxettuyen[columns_to_fix].isnull().sum().sum())
```

```python
import numpy as np

# Tính TBM1, TBM2, TBM3
for i in [1, 2, 6]:
    suffix = str(i) if i != 6 else '3'
    year = str(i)
    dulieuxettuyen[f'TBM{suffix}'] = (dulieuxettuyen[f'T{year}']*2 + dulieuxettuyen[f'L{year}'] + dulieuxettuyen[f'H{year}'] +
                                     dulieuxettuyen[f'S{year}'] + dulieuxettuyen[f'V{year}']*2 + dulieuxettuyen[f'X{year}'] +
                                     dulieuxettuyen[f'D{year}'] + dulieuxettuyen[f'N{year}']) / 10

display(dulieuxettuyen[['TBM1', 'TBM2', 'TBM3']].head())
```

```python
#Tạo biến xếp loại XL1, XL2, XL3
def xep_loai(diem):
    if diem < 5.0: return 'Y'
    elif diem < 6.5: return 'TB'
    elif diem < 8.0: return 'K'
    elif diem < 9.0: return 'G'
    else: return 'XS'

for i in [1, 2, 3]:
    dulieuxettuyen[f'XL{i}'] = dulieuxettuyen[f'TBM{i}'].apply(xep_loai)

display(dulieuxettuyen[['XL1', 'XL2', 'XL3']].head())
```

```python
# Tạo US_TBM1, US_TBM2, US_TBM3 (Min-Max Normalization sang thang điểm 4)
for i in [1, 2, 3]:
    dulieuxettuyen[f'US_TBM{i}'] = (dulieuxettuyen[f'TBM{i}'] / 10) * 4

display(dulieuxettuyen[['US_TBM1', 'US_TBM2', 'US_TBM3']].head())
```

```python
# Tạo biến KQXT
def tinh_kqxt(row):
    kt = row['KT']
    dh1, dh2, dh3 = row['DH1'], row['DH2'], row['DH3']

    if kt in ['A', 'A1']:
        diem = (dh1*2 + dh2 + dh3) / 4
    elif kt == 'B':
        diem = (dh1 + dh2*2 + dh3) / 4
    else:
        diem = (dh1 + dh2 + dh3) / 3

    return 1 if diem >= 5.0 else 0

dulieuxettuyen['KQXT'] = dulieuxettuyen.apply(tinh_kqxt, axis=1)
display(dulieuxettuyen[['KT', 'DH1', 'DH2', 'DH3', 'KQXT']].head())
```
